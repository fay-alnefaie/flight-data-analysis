```python
import pandas as pd
from clean import format_time
```

```python
df = pd.read_csv('../data/raw/flights.csv')
```

```python
df['name'] = df['name'].str.capitalize().str.strip()
df['name'].unique().tolist()
```

```python
df['name'] = df['name'].str.replace(r'\bco\.?\b','',regex= True)
df['name'] = df['name'].str.replace(r'\binc\.?\b', '', regex=True)
df['name'] = df['name'].str.replace(r'\bcorporation\.?\b','',regex=True)
df['name'] = df['name'].str.replace(r'\.$', '', regex=True)
df['name'] = df['name'].str.replace(r'air\s+lines', 'airlines', regex=True)
df['name'] = df['name'].str.replace(r'\bair\b','airlines',regex=True)
df['name'] = df['name'].replace('Virgin america', 'virgin america airlines')
df['name'] = df['name'].str.strip()
df['name'].unique().tolist()
```

```python
df['is_cancalled'] = df['dep_time'].isna().astype(int)
```

```python
df['dep_time_format'] = df['dep_time'].apply(format_time)
df[["dep_time_format","dep_time"]].head()
```

```python
df['arr_time_format'] = df['arr_time'].apply(format_time)
```

```python
df['sched_arr_time_formatted']= df['sched_arr_time'].apply(format_time)
```

```python
df['sched_dep_time_formatted'] = df['sched_dep_time'].apply(format_time)
```

```python
df['flight_date'] = pd.to_datetime(df[['year','month','day','hour', 'minute']])
df
```

```python
df['dep_time'] = pd.to_datetime(df['dep_time_format'], format='%H:%M').dt.time
```

```python
df['sched_dep_time'] = pd.to_datetime(df['sched_dep_time_formatted'], format='%H:%M').dt.time
```
